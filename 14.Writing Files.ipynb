## 14. Writing Files 

----
### 1. What is File handling in Python>

`File Handling` means working with files stored on the computer using Python:

With Python We can:

* Create a file
* Read data from a file
* Write data to a file
* Append data to a file
* Delete a file

Files are stored on secondary storage(hard disk) and help us to save data permanently.

Examples:

* Saving user data
* Storing logs
* Writing reports
* Saving configuration settings
  

## 2. Basic Steps for writing a File

When writing to a file, Python follows three main stes:

* 1. Open the file: Open()
* 2. Write data into the file.: write()
* 3. Close the file.: close()

```python
fileName = r"C:\Users\User\Desktop\data.txt"

file = open(fileName,"w")
file.write("hello world\nHey Python")
file.close()
```

### 3. File Opening Modes(Very Important)

----------

|Mode|Meaning|Description|
|----|----|----|
|`r`|Read| Open File for reading only (default)|
|`w`|Write| Creates file or overwrites exising file|
|`a`|Append|Adds data at end of the file.|
|`x`|Create| Create new file, error if exists|
|`b`|Binary|Used for images/files|
|`t`|Text| Default mode-text format|

```python
open("file.txt","r")    #-->read mode
```

```python
open("file.txt","w")   # ---> write mode only
```

```python
open("file1.txt","x")  #---> file create if exits it throw an error
```

```python
open('file1.txt',"a")   # append
```

```python
open("file.txt","t")  # text format 
```

## 4. Writing Data to a file

### Method 1: writes one string to a file.

```python
file = open(r"C:\Users\User\Desktop\data.txt","w")

file.write("Name: Rahul\n")
file.write("Age: 22\n")
file.write("Course: Python\n")
file.close()
```

```python
file = open(r"C:\Users\User\Desktop\data.txt","a")

file.write("------------------\n")
file.write("Name: Shyam\n")
file.write("Age: 25\n")
file.write("Course: C Programming\n")
file.close()
```

### 5. Writing Multiple Lines (writelines)
-----
writelines() writes multiple lines at once.

Example:

```python
file = open(r"C:\Users\User\Desktop\data.txt",'a')

lines = [
    "------------------------------\n"
    "Python programming language\n",
    "File handling\n",
    "Object Oriented Programming\n"
]
file.writelines(lines)
file.close()
```

### 6. Using "with" Statement 

Instead of manually closing fliles, Python provides with statement.

Benefits:

* Automatically closes file
* Safer
* Cleaner code

```python
fileName = r"C:\Users\User\Desktop\data.txt"

with open(fileName,"w") as file:
    file.write("Welcome to Python file Handling.\nHello World that is my second line.")
```

After the block ends ---> File automatically closes.

This is the most recommeded professionaly way.

### 7. Append Data to file 

-----
Append means add new data without deleting old data.

Mode used: "a"



```python
fileName = r"C:\Users\User\Desktop\data.txt"

with open(fileName,"a") as f:
    f.write("\n---------------------\n")
    f.write("New User Logged in\n")
```
