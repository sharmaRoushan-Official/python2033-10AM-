```python
#11. Generate a list of squares from 1 to 10 using list comprehension.

[i for i in range(1,11) if i**2]
```

```python
#12. Create a list of even numbers from 1 to 50 using list comprehension.

[i for i in range(1,51) if i%2==0]
```

```python
#13. Convert the list of strings to uppercase using list comprehension:
#    `words = ['cat', 'dog', 'lion']`

words = ['cat','dog','lion']
[w.upper() for w in words] 
```

```python
#14. Get a list of numbers divisible by both 3 and 5 from 1 to 100.

[i for i in range(1,101) if i % 3 == 0 and i % 5 == 0]
```

```python
#15. Extract all vowels from the string using list comprehension:
#    `text = "Comprehension"`

text = "Comprehension"
[ch for ch in text if ch in "aeiouAEIOU"]
```

```python
#16. Given a matrix:
#    `matrix = [[1,2,3],[4,5,6],[7,8,9]]`
#    Use list comprehension to flatten it into a single list.

matrix = [[1,2,3],[4,5,6],[7,8,9]]
[num for matrix in [[1,2,3]] for num in [1,2,3,4,5,6,7,8,9]]
```

```python
#17. Given a list of words, extract words that contain the letter ‘e’:
#    `words = ['tree', 'book', 'python', 'sky']`

words = ['tree', 'book', 'python', 'sky']
[ch for ch in words if 'e' in ch]
```

```python
#18. Create a list of all 3-digit numbers that are palindromes using list comprehension.

[i for i in range(100,1000) if str(i) == str(i)[::-1]]
```

```python
#19. From a list of numbers, replace odd numbers with `'ODD'` and keep even numbers as is:
#    `nums = [1, 2, 3, 4, 5, 6]`

nums = [1, 2, 3, 4, 5, 6]
[n if n%2==0 else 'ODD' for n in nums]
```

```python
#20. Using nested list comprehension, generate a 3x3 identity matrix.

[[1 if i == j else 0 for j in range(3)] for i in range(3)]
```
