```python
# 15. Crete a simple calulator that performs basic arithmetic operations based on two user input.
# (^,*,/,-,+)

a = int(input("Enter a first number: "))
b = int(input("Enter a second number: "))
operator = input("Enter operator(^,*,/,-,+): ")

if operator == '+':
    print(a + b)
elif operator == '-':
    print(a - b)
elif operator == '*':
    print(a * b)
elif operator == '/':
    print(a / b)
elif operator =='^':
    print(a ** b)
else:
    print("invalid operator")
```

```python
#17. Write a Python program to check whether a triangle is **equilateral**, **isosceles**, or **scalene**.

a = int(input("Enter side a: "))
b = int(input("Enter side b: "))
c = int(input("Enter side c: "))

if a == b == c:
    print("The triangle is Equilateral")
elif a == b or b == c or a == c:
    print("The triangle is Isosceles")
else:
    print("The triangle is Scalene")
```

```python
#25. Write a Python program to calculate the **electricity bill**:
#    * ≤100 units → ₹1 per unit
#    * 101–200 units → ₹1.5 per unit
#    * > 200 units → ₹2 per unit

units = int(input("Enter electricity units: "))
bill=0

if units <= 100:
    bill = units
    print("Your electricity bill is:",bill)                  
elif units>=101 and units<=200:
    bill = 100+((units-100) * 1.5)
    print("Your electricity bill is:",bill)
else:
    bill = 250+((units-200) *2)
    print("Your electricity bill is:",bill)
```
