```python
# 19. write a function prime_check(n) that returns true if number is prime.

def prime_check(n):
    if n<=1:
        return False
    for i in range(2,n): #  2,3,4
        if n%i == 0:
            return False
    return True
print(prime_check(4))
print(prime_check(7))
```

```python
# 20. Write a function to generate_primes(n) that generates all prime numbers up to n

def generate_primes(n):
    primes=[]
    for num in range(2,n+1):
        # print(num)
        is_prime = True
        for i in range(2,num):
            if num%i==0:
                is_prime = False
        if is_prime:
            primes.append(num)
    return f"List of all Primes number:{primes}"
                
generate_primes(20)
```

```python
#12. find the area of cuboid and their perimter.
def cuboid():
    length=float(input("Enter length of cuboid"))
    height=float(input("Enter height of  cuboid"))
    breadth = float(input("Enter breadth length of cuboid "))
    sur_area=2*(length*breadth+length*height+height*breadth)
    perimeter=4*(length+breadth+height)
    print(f"Area of cuboid is {sur_area}cm²")
    print(f"Perimeter of cuboid is {perimeter}cm")
cuboid()    
```

```python
#13. find the surface area of  sphare and volume
import math
def sphere():
    radis=float(input("Enter radius of sphere for Surface area and volume"))
    sur_area=4*math.pi*radis**2
    volume=(4/3)*math.pi*radis**3
    print(f"Area of sphere is {sur_area}cm²")
    print(f"Perimeter of sphere is {volume}cm³")
sphere()    
```

```python
#14. find the area of hemisphere and perimeter.
import math
def hemisphere():
    print(f"There are two types: \n 1. Curved Surface Area (only round part): \n 2. Total Surface Area (including base circle):")
    select=float(input("Select the type of area you want (Enter number 1 or 2)"))
    radis=float(input("Enter the radius value of hemisphere"))
    if select == 1:
       calc=2*math.pi*radis**2
       print(f"Curved surface area for hemisphere is {calc} cm²")
    elif select == 2:
       calc1=3*math.pi*radis**2
       print(f"Total surface area for hemisphere is {calc1} cm²")
    else :
       print("Enter proper options")
    perimeter=2*math.pi*radis
    print(f"A hemisphere doesn’t have a full “perimeter” like flat shapes, but its base is a circle\n Perimeter (circumference of base) is : {perimeter} cm")
hemisphere()
```

```python
#15. find the area of cylinder and their volume.
import math
def Cylinder():
    print(f"There are two types: \n 1. Curved Surface Area : \n 2. Total Surface Area(including top & bottom) :")
    select=float(input("Select the type of area you want (Enter number 1 or 2)"))
    radis=float(input("Enter the radius value of Cylinder"))
    height=float(input("Enter the height value of Cylinder"))
    if select == 1:
       calc=2*math.pi*radis*height
       print(f"Curved surface area for Cylinder is {calc} cm²")
    elif select == 2:
       calc1=2*math.pi*radis*(radis+height)
       print(f"Total surface area for Cylinder is {calc1} cm²")
    else :
       print("Enter proper options")
    volume=math.pi*radis**2*height
    print(f"Volume of a Cylinder is {volume}cm")
Cylinder()
```

```python
#16. find the area of cone and their volume.
import math
def Cone():
    print(f"There are two types: \n 1. Curved Surface Area : \n 2. Total Surface Area :")
    select=float(input("Select the type of area you want (Enter number 1 or 2)"))
    radis=float(input("Enter the radius value of Cone"))
    height=float(input("Enter the height value of Cone"))
    slant_height=math.sqrt(radis**2+height**2)
    if select == 1:
       calc=math.pi*radis*slant_height
       print(f"Curved surface area for Cone is {calc} cm²")
    elif select == 2:
       calc1=math.pi*radis*(radis+slant_height)
       print(f"Total surface area for Cone is {calc1} cm²")
    else :
       print("Enter proper options")
    volume=(1/3)*math.pi*radis**2*height
    print(f"Volume of a Cone is {volume}cm")
Cone()
```
