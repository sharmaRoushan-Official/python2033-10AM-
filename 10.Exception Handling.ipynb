### Basic Syntax of try-except



```python
try:
    a = 10
    b = 0
    print(a/b)  # ZeroDivisionError
except ZeroDivisionError:
    print("Cannot divide by Zero")
print("Hello world")
```

```python
a = 10
b = "a"
print(a/b)
```

```python
try:
    a = 10
    b = "a"
    print(a/b)
except TypeError:
    print("This is Type error check data type of value")
print("Done")
```

```python
try:
    a = {'a':"apple","b":"ball","c":'cat',"d":"dog"}
    print(a)
    
    print(a['e'])
except KeyError:
    print("Key is not defined in the dictionary")
```

## Handling default exception


```python
try:
    # a = 10
    # b = "a"
    # print(a/b)
    a = {'a':"apple","b":"ball","c":'cat',"d":"dog"}
    print(a)
    
    print(a['e'])
except:  # default error
    print("error")
print("Done")
```

# Multiple Exception Block

```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    print(x/y)
except ZeroDivisionError:
    print("You cannot divide anynumber by 0")
except TypeError:
    print("Invalid Input! Please Enter a valid data type.")
except ValueError:
    print("Enter a valid value.")
```

```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    print(x/y)

except (ZeroDivisionError, TypeError, ValueError):
    print("Error Occurred")
```

# Use Exceptions as Object(e)

```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    print(x/y)
except Exception as e:
    print("Error Occurred:",e)
```

### try-except-else

```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    div = x/y
except Exception as e:
    print("Error Occurred:",e)
else:
    print("Result of your Value:",div)
print("Done")
```

Notes:

if exception - > excpt runs

if no exception -> else runs

# try->except->else->finllay

```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))
    div = x/y
except Exception as e:
    print("Error Occurred:",e)
else:
    print("Result:",div)
finally:
    print("Your Code excution completed!!")
    
```

# Using raise keyword(Manually throw Exceptions)

sometimes you want ot create your own error

```python
age = 19
if age <= 18:
    raise ValueError("Age must be 18 or above")
else:
    print("You Can Vote Now")
```

```python
try:
    age = int(input("Enter Your age: "))
    if age <= 18:
        raise ValueError("Age must be greater than 18")
    else:
        print("You Can Vote Now")
except Exception as e:
    print(e)
```

```python
age = int(input("Enter Your age: "))
if age <= 17:
    raise ValueError("Age must be greater than 17")
else:
    print("You Can Vote Now")
```

```python
class InvalidAgeError(Exception):
    pass
```

```python
age = int(input("Enter your age: "))

if age<=17:
    raise InvalidAgeError("Enter your valid Age")
else:
    print("Your Age is Valid")
```

```python
class ageValidation(Exception):
    pass
```

```python
age = int(input("Enter your age: "))

if age<18:
    raise ageValidation("You are not Eligible to Vote")
else:
    print('Your Age is Valid')
```
