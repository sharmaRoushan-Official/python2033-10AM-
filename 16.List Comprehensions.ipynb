## 16. List Comprehensions

### 1. What is List Comprehensions?

---
`List comprhension` is a short and powerful way to create lists in Python using a single line of code.

Instead of writing long loops to create a list, we can write the same logic in a cleaner and faster format.

List comprhensions is compact syntax used to:

* generate a list
* transform data
* fitler element
* combine loops and condition

### 2. Why List comprehension is Used??

Benefits:

* Less code
* More readable(When use Properly)
* Faster than traditional loops
* Used heavily in real-world python programming Like: AI/ML/Deep Learning
* Best for data transformation and filtering.

### 4. General Syntax of List Comprehensions

[expression for item in iterable]


Breakdown:

----

|Part|Meaning|
|----|----|
|expresion|What you want to store in the list|
|item| Variables that takes values|
|iterable| List, tuple, set, dictionary, string, etc.|

### 10. List Comprehension with Strings

Example: Extract Vowels

```python
vowList = []
x = "Python is programming Language A"

for i in x:
    if i in "aeiouAEIOU":
        # print(i)
        vowList.append(i)
print(vowList)
```

```python
x = "Python is programming Language A"
[ch for ch in x if ch in "aeiouAEIOU"]
```

#### 11. List Comprehension with Functions

Example: Apply len() on list of words

```python
#8. Demonstrate *operator overloading* by overloading the + operator for a class Point(x, y) to add coordinates


class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    # overloading operator
    def __add__(self,other):   # value ram 
        return Point(self.x + other.x, self.y+other.y)

    def __str__(self):
        return f"{self.x},{self.y}"  # value represent
        

p1 = Point(2,3)
p2 = Point(4,5)
print(p1)
print(p2)

result = p1+p2
print("Point 1:",p1)
print("Point 2:",p2)
print("Sum:",result)
```

```python
p3 = Point(12,13)
p4 = Point(14,15)


result = p3+p4
print("Point 3:",p3)
print("Point 4:",p4)
print("Sum:",result)
```
