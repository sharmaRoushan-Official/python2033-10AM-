### 4. Decorator Concept

A decorator is basically:

* A function that takes another function
* Add extra code
* Return the modified function

```python
def addTwo():
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    result = num1+num2
    print(f"The sum of two numbers {num1} and {num2} is: {result}")

addTwo()
```

```python
def callingFunc(func):
    func()
callingFunc(addTwo)
```

```python
# modified function : added extra line of code
def decorator(func):
    def wrapper():
        print("Before Function call")
        func()
        print("After Function call")
    return wrapper
hello = decorator(addTwo)    
hello()
```

## 6. Decorator with @Syntax (Most Common)

```python
def decorator(func):
    def wrapper():
        print("Before Function Call")
        func() # Hello world  # hello()
        print("After Function call")
    return wrapper
```

```python
@decorator
def hello():
    print("Hello World")
hello()
```

```python
@decorator
def addTwo():
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    result = num1+num2
    print(f"The sum of two numbers {num1} and {num2} is: {result}")

addTwo()
```

```python
def decorator(func):
    def wrapper():
        try:
            a = int(input("Enter first number: "))
            b = int(input("Enter second number: "))
            res = a/b
        except Exception as e:
            print(e)
        else:
            print(f"Your result is:{res}")
        finally:
            print("Your Code Running Sucessfully")
        print("Before Function call")
        func() # Hello world  # hello()
        print("After Function call")
    return wrapper
```

```python
@decorator
def addTwo():
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    result = num1+num2
    print(f"The sum of two numbers {num1} and {num2} is: {result}")

addTwo()
```

```python
import datetime




@decorator
def greet():
    name = input("Enter your name:")
    print("Current time is:",datetime.datetime.now())
    print("Congratulations: ",name)
greet()
```

### 8. Decorator with *args and **kwargs



```python
def decorator1(func):
    def wrapper(*args,**kwargs):
        print("Before Function call")
        result = func(*args, **kwargs)
        print("After Function call")
        return result 
    return wrapper
```

```python
@decorator1
def add(a,b):
    return a+b
print(add(10,20))
```

### 9. Returning Value in Decorator

alway return the orginial funciton's return value

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Function is starting....")
        value = func(*args, **kwargs)
        print(f"Multiplication of two values: {value}")
        print("Function is finished....")
    return wrapper
```

```python
@decorator
def multiply(a,b):
    return a*b
multiply(3,4)
```

```python
import time
print(time.time())
```

```python
def timer(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Execution time: {end-start:.5f}")
        return result
    return wrapper
```

```python
@timer
def big_task():
    time.sleep(5)
    print("Task Done")
big_task()
```

```python
# Authentication and Authorization 

def logger(func):
    def wrapper(*args,**kwargs):
        print(f"Calling Function: {func.__name__}") # magic function __name__
        return func(*args,**kwargs)
    return wrapper
```

```python
@logger
def login(user):
    print(f"{user} logged in")
login("Yash")
```

```python
@logger
def logout(user):
    print(f"{user} Logged out")
logout("Yash")
```

```python
@logger
def greet():
    name = input("Enter your name:")
    print("Current time is:",datetime.datetime.now())
    print("Congratulations: ",name)
greet()
```
