# Pattern Practice

## Left Triangle

```python
for i in range(1,6):
    print("*"*i)
```

## Square

```python
for i in range(1,6):
    for j in range(1,6):
        print("*",end ="")
    print()
```

## Reverse Left Triangle

```python
for i in range(6,0,-1):
    print("*"*i)
```

## Increament Number Triangle

```python
for i in range(1,6):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

# Repeated number triangle

```python
for i in range(1,6):
    for j in range(1,i+1):
        print(i,end="")
    print()
```

## Reverse Number Triangle

```python
n = 5
for i in range(n,0,-1):
    for j in range(i,0,-1):
        print(j,end="")
    print()
```

## Repeated Alphabet Triangle

```python
for i in range(65,71):
    for j in range(65,i+1):
        print(chr(i),end = "")
    print()
```
