# Calculator

```python
num1 = int(input("Enter any number"))
num2 = int(input("Enter any number"))
operator = input("Enter the operator")
if operator == "+":
    print("The sum is",{num1},"and", {num2},"is",{num1+num2})
elif operator == "-":
    print("The sum is",{num1},"and", {num2},"is",{num1-num2})
elif operator == "*":
    print("The sum is",{num1},"and", {num2},"is",{num1*num2})
elif operator == "//":
     print("The sum is",{num1},"and", {num2},"is",{num1//num2})
else:
    print("Invalid Input!!")
```
