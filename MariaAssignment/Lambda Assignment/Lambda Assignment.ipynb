```python
#8) use a lambda and map to capitalize the first letter of each word in a list
word =["python","lambda","function"]
word_capital = list(map(lambda w:w.capitalize(),word))
for i in word_capital:
    print(i)
```

```python
#9) check if a string starts with "a" and count it 
words = ["apple", "banana", "avocado", "Apricot", "berry", "almond"]
count = sum(w.lower().startswith("a") for w in words)
print(count) 
```

```python
#10) use reduce() with lambda to compute the product of all elements in a list 
from functools import reduce
nums = [2,3,4,5]
reduce(lambda a,b:a*b,nums)
```
