```python
#17.  Write a program that demonstrates nested try-except blocks.
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    try:
        div = a/b
    except ZeroDivisionError:
        print("Cannot divide by zero")
except ValueError:
    print("Invalid Input!!")
else:
    print("Result:",div)    
```
