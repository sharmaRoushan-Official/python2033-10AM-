# Writing files in python 

```python
#1) r----------------- open the file for readingly .Starts reading from the beginning of life. this is default mode
file = open(r"C:\Users\DELL\Desktop\python.txt","r")
content = file.read()
print(content)
file.close()
```

```python
#2 open a file for writing only . File pointer placed at the beginning of the file .overwrites if existing file.and creates new one
filepath = (r"C:\Users\DELL\Desktop\python.txt")
file = open(filepath,"w")
file.write(" Python is easy")
file.close()
filepath = (r"C:\Users\DELL\Desktop\python.txt")
file = open(filepath,"a")
file.write("Microsoft\n")
file.write("Google\n")
file.write("Flipcart\n")
file.close()
```

```python
# Open a file in read mode
with open(r"C:\Users\DELL\Desktop\python.txt", "r") as file:
    content = file.read() 
    print(content) 
with open(r"C:\Users\DELL\Desktop\python.txt", "r") as file:
    line1 = file.readline(1) 
```

```python
# _Wb-----------------------------> Same as W but opens in binary mode.
binary_data = b"Hello World!\n This is binary data"
with open(r"C:\Users\DELL\Desktop\python.txt", "wb") as file:
    file.write(binary_data)
```
