```python
# 7) write a function that counts the vowels of string
def count_string (n):
    count= 0
    vowels = "AEIOUaeuio"
    for i in n:
        if i in vowels:
            count+=1
            print(i)
    print(count)
count_string ("favourite")
    
```

```python
#18) find the area of equilateral triangle  and their perimeter.
import math
def equilateral():
    side = float(input("Enter any side"))
    area = (math.sqrt(3)/4)*side*side
    perimeter = 3*side
    print("Area of Equilateral Triangle",area)
    print("Perimeter of Equilateral Triangle", perimeter)
equilateral()
```

```python
#22)  find the are of parallelogram and their perimter.
import math 
def parallelogram():
    base = float(input("Enter the value of base"))
    length1 = float(input("Enter the lenght1 value"))
    length2 = float(input("Enter the value of length2"))
    height = float(input("Enter the value of height"))
    area = 2*length1+length2
    perimeter = base*height
    print("Perimeter of parallelogram:",perimeter)
    print("Area of parallelogram:",area)
parallelogram()
```

```python
#23). find the are of Trapezium and their perimeter.
import math
def trapezium():
    height = float(input("Enter the height value"))
    a = float(input("Enter the 1st parallel side of trapezium"))
    b = float(input("Enter the 2nd parallel side of trapezium"))
    c = float(input("Enter the 1st non-parallel side of trapezium"))
    d = float(input("Enter the value of 2nd non-parallel side of trapezium"))
    area = 1/2*(a+b)*height
    perimeter = a+b+c+d
    print("The area of trapezium is",area)
    print("The perimeter of trapezium is",perimeter)
trapezium()
```

```python
#25) find the area of cuboid and their perimter
import math
def cuboid():
    length = float(input("Enter the value of length"))
    breadth = float(input("Enter the value of base"))
    height = float(input("Enter the value of height"))
    area = 2(length*breadth+length*height+breadth*height)
    perimeter = length+breadth+height
    print("Area of cuboid",area)
    print("Perimeter of cuboid", perimeter)
cuboid()
```
