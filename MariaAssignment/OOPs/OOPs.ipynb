```python
#1) Create a base class Animal with a method speak(). Create two subclasses Dog and Cat that override speak() and print different sounds.
class animal:
    def speak(self):
        print("Animal produces different sound")
class dog(animal):
    def speak(self):
        print("Dog barks")
class cat(animal):
    def speak(self):
        print("Meow")

c1 = cat()
c1.speak()
d1 = dog()
d1.speak()

    
```

```python
#2)Create a class Vehicle with attributes brand and model. 
#Create subclasses Car and Bike that add a method vehicle_type() returning the type of vehicle.
class vehicle:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
class car(vehicle):
    def vehicle_type(self):
        print("The brand of car is:",self.brand,"\nThe model of car is",self.model)
class bike(vehicle):
    def vehicle_type(self):
        print("The brand of bike is:",self.brand,"\nThe model of bike is",self.model)
c=car("maruti","swift")
c.vehicle_type()
b=bike("bajaj"," Pulsar ns400")
b.vehicle_type()
```

```python
#3.Create a class Person with name and age. Create a derived class Student that adds student ID and grade. Write a method to print complete details
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

class Student(Person):
    def __init__(self,Id,grade,name,age):
        super().__init__(name,age)
        self.Id=Id
        self.grade=grade 

    def details(self):
        print(f"Name : {self.name}")
        print(f"Age : {self.age}")
        print(f"ID : {self.Id}")
        print(f"Grade : {self.grade}")
        
        

s=Student("100","Grade 11","Maria","15")

s.details()
```

```python
#4.Demonstrate *multilevel inheritance* using classes A → B → C, each with a method. Call all three methods from an object of class C.
class A:
    def agent(self):
        print("Agent should be patriot to the nation")
class B(A):
    def requirements(self):
        print("He should be intellegent and physically fit ")
class C(B):
    def work(self):
        print("To collect intel and report to officers")

c=C()
c.agent()
c.requirements()
c.work()
```

```python
#5.Create a parent class Shape with a method area(). Derive classes Circle and Rectangle. Implement the area method in each subclass.
import math
class Shape:
    def area(self):
        print("Area of shape")
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        print("Area of Circle =", math.pi * self.radius * self.radius)

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        print("Area of Rectangle =", self.length * self.width)

c = Circle(5)
r = Rectangle(4, 6)
c.area()
r.area()
```
