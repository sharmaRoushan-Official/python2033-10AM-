```python
num1 = int(input("Enter a first number: "))
num2 = int(input("Enter a second number: "))
op = input("Enter Operator  +,-,*,/: ")
print(op)
if op == '+':
    sum1 = num1+num2
    print(f"Sum of two Numbers is:{sum1}")
elif op == '-':
    print(num1 - num2)
elif op == '*':
    print(num1 * num2)
elif op == '/':
    print(num1 / num2)
elif op =='^':
    print(num1 ** num2)
else:
    print("invalid operator")
```
