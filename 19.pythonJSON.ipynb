### 5. Convert Python Dictionary to JSON [Serialization]

This is called Serialization


json.dumps() ---> Convert Python Object to JSON string

```python
data = {
    "name":"Amit",
    "age":30,
    'city':"Hydrabad"
}
print(data,type(data)) # python - dictionary: single quotes
print(data['name'])
print(data['age'])
print(data['city'])
```

```python
json_data = json.dumps(data) # data- python objects-[dictionary]
print(json_data)   # jsons---> double quotes
```

```python
json_data['name']
```

## Note: 
#### So, dumps() return a JSON String.

## 6. Convert JSON to Python dictionary (DeSerialization)

This is called Deserialization.

json.loads() --> Convert JSON string to Python Object 

```python
python_data = json.loads(json_data)  # json_data: json_string
print(python_data)
```

```python
python_data['name'] ="Rudra"
python_data
```

```python
string_data = json.dumps(python_data)
string_data
```

```python
json_string = '{"name":"Yash","age":35,"city":"Pune"}'
json_string
```

```python
json_string['name']
```

```python
python_data = json.loads(json_string)
print(python_data)  # python Object 
```

loads(): ---> json string to python dictionary[object]  --> modifictionary or change data 

dumps(): ---> python dictionary to json string ---> readable for browser. string not changeable or readable for brwoser.

### 15. JSON Encoding Problem

(Non-Serializable Object)

some objects cannot be conversted directly to JSON.

```python
from datetime import datetime
```

```python
print(datetime.now())
```

```python
data = {'time': datetime.now()}

print(json.dumps(data))
```

### solution : Convert datetime to string

```python
data = {'time':str(datetime.now())}
print(json.dumps(data))
print(type(json.dumps(data)))
```

### 17. Real-Time Example (API Response JSON)

```python
{
    "status":"Success",
    "data":[
        {"id":1,"product":"Laptop"},
        {"id":2,"product":"Mobile"}
    ]
}
```

Python Access:

```python
response = {
    "status":"Success",
    "data":[
        {"id":1,"product":"Laptop"},
        {"id":2,"product":"Mobile"}
    ]
}
print(response)
```

```python
for item in response['data']:
    print(item["id"])
    print(item['product'])
```
