```python
### 19. write a function prime_check(n) that returns true if number is prime.

def prime_check(n):
    if n<=1:
        return False
    for i in range(2,n): #  2,3,4
        if n%i == 0:
            return False
    return True
print(prime_check(4))
print(prime_check(7))
```

```python
## 20. Write a function to generate_primes(n) that generates all prime numbers up to n

def generate_primes(n):
    primes=[]
    for num in range(2,n+1):
        # print(num)
        is_prime = True
        for i in range(2,num):
            if num%i==0:
                is_prime = False
        if is_prime:
            primes.append(num)
    return f"List of all Primes number:{primes}"
                
generate_primes(20)
```
