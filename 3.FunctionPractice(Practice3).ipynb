```python
# 6. wirte a function max_of_three(a,b,c) that return maximum of three numbers

def max_of_three(a,b,c):
    if a>b and a>c:
        return f"Greatest number is:{a}"
    elif b>a and b>c:
        return f"Greatest number is:{b}"
    elif c>a and c>b:
        return f"Greatest number is:{c}"
    elif a == b and a>c:  # a= 10, b=10, c=11
        return f"Greatest number is:{a}"
    elif b == c and b>a:
        return f"Greatest number is:{b}"
    elif a == c and c>b:
        return f"Greatest number is:{c}"

max_of_three(10,11,78)
```

```python
## 9. Write a function count_vowel(s)that counts vowels in a string.

string = "Python is a programming language"

vow= 'aeiouAEIOU'
vowChar = []
countVow = 0
for i in string:
    if i in vow:
        # print(i)
        vowChar.append(i)
        countVow+=1
print("Vowel Characters in your string is:",vowChar,"\nTotal Vowel Value:",countVow)
```

```python
## 11. Write a function is_palindrome(s) that checks whether the string is palindrome or not

# "mom"
# "dad"
# "kayak"
# "madam"
# "mam"
# "pop"
# "nayan"
# "malayalam"

word = "Mom"

wordLower = word.lower()

# print("original Word:",word)
rev_word = ""
for i in word:
    rev_word = i.lower() + rev_word
    # print(i)
# print("Rersed Word:",rev_word)


if wordLower == rev_word:
    print("Palindrome Word:",word)
else:
    print("Not Palindrome word:",word)
```
