```python
# 24. Print all duplicate elements in a list.

numbers = [1, 2, 3, 2, 4, 5, 1, 6, 3]

for i in range(len(numbers)):
    count = 0

    for j in range(len(numbers)):
        if numbers[i] == numbers[j]:
            count += 1

    if count > 1:
        duplicate = False

        for k in range(i):
            if numbers[k] == numbers[i]:
                duplicate = True
                break

        if not duplicate:
            print(numbers[i])
```

```python
#38. Find whether a number is Armstrong number or not.

num = int(input("Enter a number: "))
temp = num
digits = len(str(num))
total = 0

while num > 0:
    digit = num % 10
    total += digit ** digits
    num //= 10

if total == temp:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
#39. Print all Armstrong numbers between 1 and 500

num = 1

while num <= 500:
    temp = num
    total = 0
    digits = len(str(num))

    while temp > 0:
        digit = temp % 10
        total += digit ** digits
        temp //= 10

    if total == num:
        print(num)

    num += 1
```

```python
#40. Check whether a number is prime or not.

num = int(input("Enter a number: "))
i = 2
prime = True

while i < num:
    if num % i == 0:
        prime = False
        break
    i += 1

if prime and num > 1:
    print("Prime")
else:
    print("Not Prime")
```

```python
#42. Find the greatest common divisor (GCD) using while loop.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    a, b = b, a % b

print("GCD =", a)
```

```python
#43. Find the least common multiple (LCM) using while loop.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

greater = max(a, b)

while True:
    if greater % a == 0 and greater % b == 0:
        print("LCM =", greater)
        break
    greater += 1
```
