```python
#Q10. Create a function to count the number of vowels and consonants in a string. 
#String = “The quick brown fox jumps over the lazy dog” 


string = "The quick brown fox jumps over the lazy dog"
def vc_count(string):
    vowels = "aeiouAEIOU"
    vowelcount = 0
    consonantcount = 0

    for char in string:
        if char.alphabate():  
            if char in vowels:
                vowelcount += 1
            else:
                consonantcount += 1

    return vowelcount, consonantcount

vowels, consonants = vc_count(string)
print("Numbers of vowel:", vowels)
print("Numbers of consonant:", consonants)
```
