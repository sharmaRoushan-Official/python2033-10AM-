### 2.🎯 Number Guessing Game

```python
from IPython.display import HTML, display

import random

print("===== Game Khelo Tukka Lagaike!!! =====")

gnumber = random.randint(1, 100)

while True:
    try:
        guess = int(input("Enter your guess (1-100): "))

        if guess < gnumber:
            print("Bahote niche gaya re tu")
            display(HTML("""
            <iframe width="400" height="700"
            src="https://www.youtube.com/embed/Z-hFKtNTHgg?autoplay=1&mute=0"
            frameborder="0"
            allow="autoplay">
            </iframe>
            """))

        elif guess > gnumber:
            print("jaada ho gaya re")
            display(HTML("""
            <iframe width="400" height="700"
            src="https://www.youtube.com/embed/P9rkprWACdo?autoplay=1&mute=0"
            frameborder="0"
            allow="autoplay">
            </iframe>
            """))

        else:
            print("Sabase Bada Sattebaaz!")
            print("satte ki rashi:", gnumber)
            display(HTML("""
            <iframe width="400" height="700"
            src="https://www.youtube.com/embed/4Cduq0z0tZQ?autoplay=1&mute=0"
            frameborder="0"
            allow="autoplay">
            </iframe>
            """))
            break

    except ValueError:
        print("Jis raah pe tum chal rahe hona beta bure faso ge")
        display(HTML("""
            <iframe width="700" height="400"
            src="https://www.youtube.com/embed/8pHKxQ_Ogx0?autoplay=1&mute=1"
            frameborder="0"
            allow="autoplay">
            </iframe>
            """))
        display(HTML("""
            <iframe width="700" height="400"
            src="https://www.youtube.com/embed/Z9oThmTT6j0?autoplay=1&mute=0"
            frameborder="0"
            allow="autoplay">
            </iframe>
            """))
```
