```python
#1 Wrp to find the area of square and thier perimeter of square using class and object.

class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

    def perimeter(self):
        return 4 * self.side

side = float(input("Enter side: "))
s = Square(side)

print("Area =", s.area())
print("Perimeter =", s.perimeter())
```

```python
#2. wap to find area of rectangle and thier perimeter.

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

l = float(input("Enter length: "))
w = float(input("Enter width: "))

r = Rectangle(l, w)

print("Area =", r.area())
print("Perimeter =", r.perimeter())
```

```python
#3. wap to find area of circle and their circumference and their diameter.

import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def circumference(self):
        return 2 * math.pi * self.radius

    def diameter(self):
        return 2 * self.radius

r = float(input("Enter radius: "))

c = Circle(r)

print("Area =", c.area())
print("Circumference =", c.circumference())
print("Diameter =", c.diameter())
```

```python
#4. find the area scalen triangle and their perimeter.

import math

class ScaleneTriangle:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def area(self):
        s = (self.a + self.b + self.c) / 2
        return math.sqrt(s * (s-self.a) * (s-self.b) * (s-self.c))

    def perimeter(self):
        return self.a + self.b + self.c

a = float(input("Side 1: "))
b = float(input("Side 2: "))
c = float(input("Side 3: "))

t = ScaleneTriangle(a, b, c)

print("Area =", t.area())
print("Perimeter =", t.perimeter())
```

```python
#5. find the area of equilateral triangle  and their perimeter.

class EquilateralTriangle:

    def getData(self):
        self.side = float(input("Enter Side : "))

    def area(self):
        print("Area =", (1.732 / 4) * self.side * self.side)

    def perimeter(self):
        print("Perimeter =", 3 * self.side)

obj = EquilateralTriangle()
obj.getData()
obj.area()
obj.perimeter()
```

```python
#6. find the area of right-angle triangle and thier perimeter.

class RightTriangle:

    def getData(self):
        self.base = float(input("Enter Base : "))
        self.height = float(input("Enter Height : "))
        self.hypotenuse = float(input("Enter Hypotenuse : "))

    def area(self):
        print("Area =", (self.base * self.height) / 2)

    def perimeter(self):
        print("Perimeter =", self.base + self.height + self.hypotenuse)

obj = RightTriangle()
obj.getData()
obj.area()
obj.perimeter()
```

```python
#12. find the area of cuboid and their perimter.

class Cuboid:

    def getData(self):
        self.length = float(input("Enter Length : "))
        self.width = float(input("Enter Width : "))
        self.height = float(input("Enter Height : "))

    def area(self):
        print("Surface Area =", 2 * ((self.length * self.width) + (self.length * self.height) + (self.width * self.height)))

    def perimeter(self):
        print("Perimeter =", 4 * (self.length + self.width + self.height))

obj = Cuboid()
obj.getData()
obj.area()
obj.perimeter()
```

```python
#13. find the surface area of  sphare and volume.

class Sphere:

    def getData(self):
        self.radius = float(input("Enter Radius : "))

    def area(self):
        print("Surface Area =", 4 * 3.14 * self.radius * self.radius)

    def volume(self):
        print("Volume =", (4/3) * 3.14 * self.radius * self.radius * self.radius)

obj = Sphere()
obj.getData()
obj.area()
obj.volume()
```

```python
#15. find the area of cylinder and their volume.

class Cylinder:

    def getData(self):
        self.radius = float(input("Enter Radius : "))
        self.height = float(input("Enter Height : "))

    def area(self):
        print("Surface Area =", 2 * 3.14 * self.radius * (self.radius + self.height))

    def volume(self):
        print("Volume =", 3.14 * self.radius * self.radius * self.height)

obj = Cylinder()
obj.getData()
obj.area()
obj.volume()
```

```python
#16. find the area of cone and their volume.

class Cone:

    def getData(self):
        self.radius = float(input("Enter Radius : "))
        self.height = float(input("Enter Height : "))
        self.slant = float(input("Enter Slant Height : "))

    def area(self):
        print("Surface Area =", 3.14 * self.radius * (self.radius + self.slant))

    def volume(self):
        print("Volume =", (3.14 * self.radius * self.radius * self.height) / 3)

obj = Cone()
obj.getData()
obj.area()
obj.volume()
```
