```python
## ✅ *20 Python OOP Practice Questions*

### *Inheritance (1–5)*

#1. Create a base class Animal with a method speak(). Create two subclasses Dog and Cat that override speak() and print different sounds.

class Animal:

    def speak(self):
        print("Animal Sound")


class Dog(Animal):

    def speak(self):
        print("Dog Barks")


class Cat(Animal):

    def speak(self):
        print("Cat Meows")


d = Dog()
c = Cat()

d.speak()
c.speak()
```

```python
#2. Create a class Vehicle with attributes brand and model. Create subclasses Car and Bike that add a method vehicle_type() returning the type of vehicle.

class Vehicle:

    def __init__(self, brand, model):
        self.brand = brand
        self.model = model


class Car(Vehicle):

    def vehicle_type(self):
        print("Vehicle Type : Car")
        print("Brand :", self.brand)
        print("Model :", self.model)


class Bike(Vehicle):

    def vehicle_type(self):
        print("Vehicle Type : Bike")
        print("Brand :", self.brand)
        print("Model :", self.model)


c = Car("Toyota", "Fortuner")
b = Bike("Honda", "Shine")

c.vehicle_type()
b.vehicle_type()
```

```python
#3. Create a class Person with name and age. Create a derived class Student that adds student ID and grade. Write a method to print complete details.

class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):

    def __init__(self, name, age, sid, grade):
        super().__init__(name, age)
        self.sid = sid
        self.grade = grade

    def display(self):
        print("Name :", self.name)
        print("Age :", self.age)
        print("Student ID :", self.sid)
        print("Grade :", self.grade)


s = Student("Rudra", 21, 101, "A")

s.display()
```

```python
#5. Create a parent class Shape with a method area(). Derive classes Circle and Rectangle. Implement the area method in each subclass.

class Shape:

    def area(self):
        pass


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print("Area =", 3.14 * self.radius * self.radius)


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        print("Area =", self.length * self.width)


c = Circle(5)
r = Rectangle(10, 5)

c.area()
r.area()
```

```python
#8. Demonstrate *operator overloading* by overloading the + operator for a class Point(x, y) to add coordinates.

class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):

        x = self.x + other.x
        y = self.y + other.y

        return Point(x, y)

    def display(self):
        print("X =", self.x)
        print("Y =", self.y)


p1 = Point(10, 20)
p2 = Point(30, 40)

p3 = p1 + p2

p3.display()
```

```python
### *Encapsulation (11–15)*

#11. Create a class Account with private attributes __balance. Provide methods deposit() and withdraw() to modify it safely.

class Account:

    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        self.__balance = self.__balance + amount
        print("Balance =", self.__balance)

    def withdraw(self, amount):

        if amount <= self.__balance:
            self.__balance = self.__balance - amount
            print("Balance =", self.__balance)

        else:
            print("Insufficient Balance")


obj = Account()

obj.deposit(5000)
obj.withdraw(2000)
```

```python
### *Data Abstraction (16–20)*

#16. Create an abstract class Vehicle with an abstract method start(). Implement it in subclasses Car and Bike.

from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car Started")


class Bike(Vehicle):

    def start(self):
        print("Bike Started")


c = Car()
b = Bike()

c.start()
b.start()
```

```python
#17. Create an abstract class Shape with abstract methods area() and perimeter().

from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):

    def __init__(self, r):
        self.r = r

    def area(self):
        print("Area =", 3.14 * self.r * self.r)

    def perimeter(self):
        print("Circumference =", 2 * 3.14 * self.r)


class Square(Shape):

    def __init__(self, side):
        self.side = side

    def area(self):
        print("Area =", self.side * self.side)

    def perimeter(self):
        print("Perimeter =", 4 * self.side)


c = Circle(5)
s = Square(4)

c.area()
c.perimeter()

s.area()
s.perimeter()
```

```python
#18. Define an abstract class FileHandler with methods read() and write(). Create child classes TextFileHandler and BinaryFileHandler.

from abc import ABC, abstractmethod

class FileHandler(ABC):

    @abstractmethod
    def read(self):
        pass

    @abstractmethod
    def write(self):
        pass


class TextFileHandler(FileHandler):

    def read(self):
        print("Reading Text File")

    def write(self):
        print("Writing Text File")


class BinaryFileHandler(FileHandler):

    def read(self):
        print("Reading Binary File")

    def write(self):
        print("Writing Binary File")


t = TextFileHandler()
b = BinaryFileHandler()

t.read()
t.write()

b.read()
b.write()
```

```python
#19. Create an interface-like abstract class Remote with methods power_on() and power_off(). Implement it for TV and AC classes.

from abc import ABC, abstractmethod

class Remote(ABC):

    @abstractmethod
    def power_on(self):
        pass

    @abstractmethod
    def power_off(self):
        pass


class TV(Remote):

    def power_on(self):
        print("TV ON")

    def power_off(self):
        print("TV OFF")


class AC(Remote):

    def power_on(self):
        print("AC ON")

    def power_off(self):
        print("AC OFF")


tv = TV()
ac = AC()

tv.power_on()
tv.power_off()

ac.power_on()
ac.power_off()
```

```python
#20. Implement an abstract class Payment with method make_payment(). Create subclasses CardPayment, UPIPayment, and WalletPayment with different implementations.

from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def make_payment(self):
        pass


class CardPayment(Payment):

    def make_payment(self):
        print("Payment by Card")


class UPIPayment(Payment):

    def make_payment(self):
        print("Payment by UPI")


class WalletPayment(Payment):

    def make_payment(self):
        print("Payment by Wallet")


c = CardPayment()
u = UPIPayment()
w = WalletPayment()

c.make_payment()
u.make_payment()
w.make_payment()
```
