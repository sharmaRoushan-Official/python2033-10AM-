# 14. Writing files in Python

 * _To Write to a text file in python, you follow these steps:_* _First, Open the text file for writing(appending) using open() functions._
 * _Second, write to the text file using the write() or writelines() method._
 * _Third, close the file using close() method._
* _Characters---------------------> Functions._
 * _W -----------------------------> Open file for writing only. File pointer placed at beginning of the file. Overwrites if existing file and Creates a new one if it does not exists._
 
 * _r------------------------------> Open the file for reading only. Starts reading from the beginning of the file.This is default mode._
 
 * _rb-----------------------------> Open a file for reading only. Starts reading from beginning of the file._
 
 * _r+-----------------------------> Open file for reading only. File pointer placed at beginning of the file._
 
 * _Wb-----------------------------> Same as W but opens in binary mode._
 
 * _W+-----------------------------> Same as W but open allows to read from the file._
 
 * _Wb+----------------------------> Open a file for appending. Starts writing at the end of the file. Creates a new file if file does not exist._
 
 * _ab-----------------------------> Same as a but in binary format. Creates a new file if file does not exist._
 
 * _ab+----------------------------> Same as ab but also open for reading._



```python
#1.write()

file = open("Rudra22.txt", "w")

file.write("Hello Python")

file.close()

print("Data Written Successfully")
```

```python
#2.append 
file = open("Rudra22", "a")

file.write("\nHeHe Boyee")

file.close()

print("Data Appended Successfully")
```

```python
#3.read
file = open("Rudra22.txt", "r")

print(file.read())

file.close()
```

```python
#1.write()

file = open("Rudra4.txt", "w")

file.write("Python only")

file.close()
```

```python
#2.writelines()

file = open("Rudra4.txt", "w")

data = ["Apple\n", "Mango\n", "Orange\n"]

file.writelines(data)

file.close()
```

```python
#3.read line

file = open("Rudra4.txt", "r")

for line in file:
    print(line)

file.close()
```
