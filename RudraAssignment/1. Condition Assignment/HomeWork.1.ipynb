```python
# 15. Crete a simple calulator that performs basic arithmetic operations based on two user input.
# (^,*,/,-,+)

def calculator():
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))
    op = input("Enter operator (+, -, *, /, ^): ")

    if op == "+":
        print("Result =", a + b)

    elif op == "-":
        print("Result =", a - b)

    elif op == "*":
        print("Result =", a * b)

    elif op == "/":
        if b != 0:
            print("Result =", a / b)
        else:
            print("Division by zero ❌")

    elif op == "^":
        print("Result =", a ** b)

    else:
        print("Invalid operator ❌")
        

# function call
calculator()
```

```python
### 🚀 *Advanced / Real-Life Scenarios (21–30)*

#21. Write a Python program to calculate *discount* based on purchase amount:

amount = int(input("Enter the purchase amount: $"))

discount = 0.0


if amount > 1000:
    discount = 0.20  
elif amount >= 500:  
    discount = 0.10  
else:
    discount = 0.0   

discount_amount = amount * discount
final_amount = amount - discount_amount

print("\n--- Bill Details ---")
print("Original Amount: $" + str(amount))

if discount > 0:
    print("Discount Applied: " + str(discount * 100) + "%")
    print("Discount Amount: -$" + str(discount_amount))
else:
    print("No Discount Applied.")

print("Final Amount to Pay: $" + str(final_amount))
print("--------------------")
```
