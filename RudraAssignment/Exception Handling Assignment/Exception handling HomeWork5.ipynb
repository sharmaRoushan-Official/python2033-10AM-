```python
# ✅ *20 Python Try–Except–Else–Finally Practice Questions*

### *1. Basic Syntax*

#Write a Python program that takes a number from the user and prints its square. Use try and except to handle invalid (non-numeric) input.

try:
    num = int(input("Enter Number : "))
    print("Square =", num * num)

except ValueError:
    print("aBE SALE")
```

```python
### *2. Multiple Except Blocks*

#Write code that divides two numbers entered by the user. Handle:

#* ValueError (invalid input)
#* ZeroDivisionError (denominator = 0)

try:
    num1 = int(input("Enter First Number : "))
    num2 = int(input("Enter Second Number : "))

    print("Division =", num1 / num2)

except ValueError:
    print("Invalid Input")

except ZeroDivisionError:
    print("Cannot Divide By Zero")
```

```python
### *3. Using Else Block*

#Write a program where:

#* try block attempts to open a file data.txt
#* except handles FileNotFoundError
#* else prints “File opened successfully”

try:
    file = open("Rudra.txt", "r")

except FileNotFoundError:
    print("File Not Found")

else:
    print("File Opened Successfully")
    file.close()
```

```python
### *4. Using Finally Block*

#Write a program that tries to read a file. Regardless of success or failure, print “Operation completed” using finally.

try:
    file = open("D:Rudra.txt", "r")
    print(file.read())

except FileNotFoundError:
    print("File Not Found")

finally:
    print("Operation Completed")
```

```python
### *5. Catching Multiple Exceptions Together*

# Write code to convert a user’s input to an integer and catch *ValueError and TypeError* in a single except block.

try:
    num = int(input("Enter Number : "))
    print(num)

except (ValueError, TypeError):
    print("Tu thodasa pagal he kya")
```

```python
### *6. Raising Exception*

#Write code that:

#* Asks the user for age
#* If age < 0, raise a ValueError
#* Use try-except to handle the raised exception

try:
    age = int(input("Enter Age : "))

    if age < 0:
        raise ValueError

    print("Age =", age)

except ValueError:
    print("Age galat he")
```

```python
### *7. Nested Try-Except*

#Simulate:

#* Outer try: Convert input to integer
#* Inner try: Divide 100 by that number
#  Handle both errors separately.

try:

    num = int(input("Enter Number : "))

    try:
        print("Answer =", 100 / num)

    except ZeroDivisionError:
        print("Cannot Divide By Zero")

except ValueError:
    print("galat likhata he")
```

```python
### *8. Try–Except–Finally With File Closing*

#Open a file, read its contents, and ensure it closes in finally block even if an error occurs.

try:
    file = open("Rudra.txt", "r")
    print(file.read())

except FileNotFoundError:
    print("File Not Found")

finally:
    try:
        file.close()
    except:
        pass

    print("File Closed")
```

```python
### *9. Try With Return Statement*

#Write a function using try-except-finally where the try returns a value but the finally also contains a return. Observe the output.

def demo():

    try:
        return 10

    except:
        return 20

    finally:
        return 30

print(demo())
```

```python
### *10. Custom Exception*

#Create a custom exception InvalidMarkError and raise it when marks > 100 or < 0. Handle it using try-except.

class InvalidMarkError(Exception):
    pass

try:

    marks = int(input("Enter Marks : "))

    if marks < 0 or marks > 100:
        raise InvalidMarkError

    print("Marks =", marks)

except InvalidMarkError:
    print("Tereko marks chahiye ya nahi")
```

```python
### *11. Else With No Exception*

#Write a program where:

#* Try converts input to float
#* Except handles invalid input
#* Else prints “Conversion successful”

try:
    num = float(input("Enter Number : "))

except ValueError:
    print("Invalid Input")

else:
    print("Conversion Successful")
    print("Number =", num)
```

```python
### *12. Handling KeyError*

#Write a program that accesses a dictionary key entered by the user and handles KeyError.

student = {
    "Name": "Rahul",
    "Age": 20,
    "City": "Kolhapur"
}

try:
    key = input("Enter Key : ")

    print(student[key])

except KeyError:
    print("Key Not Found")
```

```python
### *13. Handling IndexError*

#Ask the user for a list index and handle out-of-range cases.

numbers = [10, 20, 30, 40, 50]

try:
    index = int(input("Enter Index : "))

    print(numbers[index])

except IndexError:
    print("Index Out Of Range")

except ValueError:
    print("Invalid Input")
```

```python
### *14. Converting to Integer in a Loop*

#Loop until the user enters a valid integer. Use try-except to prevent program crash.

while True:

    try:
        num = int(input("Enter Integer : "))
        print("Number =", num)
        break

    except ValueError:
        print("Invalid Input")
```

```python
### *15. Try-Except Inside a Function*

#Create a function that calculates the square root of a number. Handle ValueError if input < 0.

import math

def square_root(num):

    try:

        if num < 0:
            raise ValueError

        print("Square Root =", math.sqrt(num))

    except ValueError:
        print("Negative Number Not Allowed")

num = int(input("Enter Number : "))

square_root(num)
```

```python
### *16. Reading From a Possibly Corrupt File*

#Write a program that tries to read and convert each line of a file into an integer. Handle:

#* FileNotFoundError
#* ValueError (line cannot convert)

try:

    file = open("Rudra.txt", "r")

    for line in file:

        try:
            print(int(line))

        except ValueError:
            print("Invalid Number :", line)

    file.close()

except FileNotFoundError:
    print("File Not Found")
```

```python
### *17. Finally for Cleanup*

#Simulate allocating a resource (print “Resource allocated”) and always free it in finally (“Resource released”) even if an exception occurs.

try:

    print("Resource Allocated")

    num = 10 / 2

except Exception:
    print("Error")

finally:
    print("Resource Released")
```

```python
### *18. Multiple Excepts + Else + Finally*

#Solve this pattern:


#try:
    # risky operation
#except ValueError:
    # handle value error
#except TypeError:
    # handle type error
#else:
    # success
#finally:
    # always executed


#Write any valid code to demonstrate this structure.

try:

    num = int(input("Enter Number : "))

    print("Value =", num)

except ValueError:
    print("Value Error")

except TypeError:
    print("Type Error")

else:
    print("No Exception")

finally:
    print("Program Finished")
```

```python
### *19. Re-Raising Exceptions*

#Catch a ValueError, print a message, and re-raise the exception.

try:

    num = int(input("Enter Number : "))

    if num < 0:
        raise ValueError("Negative Number")

except ValueError as e:

    print("Error :", e)

    raise
```

```python
### *20. Handling Unexpected Exceptions*

#Use a generic exception handler except Exception as e: to print error type and message.

try:

    num1 = int(input("Enter First Number : "))
    num2 = int(input("Enter Second Number : "))

    print("Answer =", num1 / num2)

except Exception as e:

    print("Error Type :", type(e).__name__)
    print("Error Message :", e)
```
