```python
#7. find the are of rhombus and thier perimeter.
import math
def rhombus():
    print(f" Please Select the Known variable type for rhombus to find area and parameter from given list \n 1 : Using Diagonal \n 2 : Using side length and interior angle \n 3 : Using base and height (but cannot derive parameter) ")
    gear=float(input("Enter the option number From above given list"))
    if gear==1:
        diag1=float(input("Enter diagonal 1 value"))
        diag2=float(input("Enter diagonal 2 value"))
        newdiag=(1/2)*diag1*diag2
        peri1=2*math.sqrt((diag1*diag1)+(diag2*diag2))
        print(f"The Area of Rhombus is {newdiag} cm²")
        print(f"The Perimeter of Rhombus is {peri1} cm")
    elif gear==2:
        side=float(input("Enter side value"))
        inter=float(input("Enter interior angle value"))
        newform=side*side*math.sin(math.radians(inter))
        peri2=4*side
        print(f"The Area of Rhombus is {newform} cm²")
        print(f"The Perimeter of Rhombus is {peri2} cm")
    elif gear==3:    
        heig=float(input("Enter height value"))
        base=float(input("Enter base value"))
        area1=heig*base
        print(f"The Area of Rhombus is {area1} cm²")
    else:
        print("Invalid option selected.")    
rhombus()    
```

```python
#8. find the area of isosclaes triangle and their perimeter.
import math
def isoceles():
    print(f" Please Select the Known variable type for isoceles trianle to find area and parameter from given list \n 1 : Using Base and Height \n 2 :Using Three Sides \n 3 : Using Two Sides and an Angle ")
    gear=float(input("Enter the option number From above given list"))
    if gear==1:
        base1=float(input("Enter base value"))
        heig2=float(input("Enter height value"))
        area1=(1/2)*base1*heig2
        peri1=base1+2*math.sqrt(heig2**2+(base1/2)**2)
        print(f"The Area of isoceles is {area1} cm²")
        print(f"The Perimeter of isoceles is {peri1} cm")
    elif gear==2:
        side=float(input("Enter side If only the equal side length  are known"))
        side1=float(input("Enter base value"))
        newform=(side1/2)*math.sqrt(side**2-(side1**2/4))
        peri2=2*side+side1
        print(f"The Area of isosceles is {newform} cm²")
        print(f"The Perimeter of isosceles is {peri2} cm")
    elif gear==3:    
        side1=float(input("Enter two equal side  value"))
        angel=float(input("Enter Angle "))
        area4=(1/2)*side1**2*math.sin(math.radians(angel))
        peri3=2*side1*(1+math.sin(math.radians(angel/2)))
        print(f"The Area of isosceles is {area4} cm²")
        print(f"The Perimeter of isosceles is {peri3} cm")
    else:
        print("Invalid option selected.")    
isoceles()    
```

```python
#15. find the area of cylinder and their volume.
import math
def Cylinder():
    print(f"There are two types: \n 1. Curved Surface Area : \n 2. Total Surface Area(including top & bottom) :")
    select=float(input("Select the type of area you want (Enter number 1 or 2)"))
    radis=float(input("Enter the radius value of Cylinder"))
    height=float(input("Enter the height value of Cylinder"))
    if select == 1:
       calc=2*math.pi*radis*height
       print(f"Curved surface area for Cylinder is {calc} cm²")
    elif select == 2:
       calc1=2*math.pi*radis*(radis+height)
       print(f"Total surface area for Cylinder is {calc1} cm²")
    else :
       print("Enter proper options")
    volume=math.pi*radis**2*height
    print(f"Volume of a Cylinder is {volume}cm")
Cylinder()
```

```python
#16. find the area of cone and their volume.
import math
def Cone():
    print(f"There are two types: \n 1. Curved Surface Area : \n 2. Total Surface Area :")
    select=float(input("Select the type of area you want (Enter number 1 or 2)"))
    radis=float(input("Enter the radius value of Cone"))
    height=float(input("Enter the height value of Cone"))
    slant_height=math.sqrt(radis**2+height**2)
    if select == 1:
       calc=math.pi*radis*slant_height
       print(f"Curved surface area for Cone is {calc} cm²")
    elif select == 2:
       calc1=math.pi*radis*(radis+slant_height)
       print(f"Total surface area for Cone is {calc1} cm²")
    else :
       print("Enter proper options")
    volume=(1/3)*math.pi*radis**2*height
    print(f"Volume of a Cone is {volume}cm")
Cone()
```

```python
#Write a function prime_check(n) that returns True if number is prime
def prime_check():
    num=int(input("Enter any number"))
    if num<=1:
        print("Entered number is not valid")
    elif num>=1 :
        for i in range(2,num):
           if num%i==0:
               print(num, "is not a prime number")
               return
        print(num, "is a prime number")
prime_check()            
            
        
```
