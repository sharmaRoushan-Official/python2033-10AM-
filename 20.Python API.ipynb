### 5. Your First Python API Program (GET Request)

URL- Uniform Resourse Locator

https://jsonplaceholder.typicode.com/users

https://jsonplaceholder.typicode.com/users

code (With Line-by-line)


```python
# step 1: Import Requests Library

import requests
```

Note:

This library allows Python to send HTTP requests

```python
# step 2: Store API URL


url = "https://jsonplaceholder.typicode.com/users"
url
```

This is the API endpoints(Server address)

```python
# step3: Send GET request ---> Read data from server

response = requests.get(url)  # GET ---> Reads data from server
print(response)   # if you will get 200 then server response successfully 
```

Note:

* Sends reques to server
* Server replies and stores result in response variable

```python
# step 4 : check status code 

print(response.status_code)
```

Note:

* 200 ---> Success
* 404 ---> Not Found
* 500 ---> Server Error

```python
# step 5: Convert response to JSON

data = response.json()   # string to object 
print(data)
```

step 1: import API

setp 4: response ---> 200 port number Success

### 8. POST Request (Send Data to API)

Create a new user

```python
# import requests 

url = "https://jsonplaceholder.typicode.com/posts"  # endpoint
url
```

```python
payload = {
    "title":"Python API Training",
    "body":"Python is a programming Language",
    "userId":101
}
payload
```

Note:

payload = data we send to server

```python
response = requests.post(url,json=payload)  # convert python object to json string ---> dumps()
print(response)
```

##### The HTTP 201 Created status code is a success response indicating that a new 
##### resource has been successfully created on the server.

Note:

* post() ---> create data in server
* json= payload() ---> auotmatically convertys to JSON()

```python
print(response.status_code)
```

```python
print(response.json())  # convert json_string to python object 
```

#### 9. PUT Request (Updata Data) ---> Major data

```python
url = "https://jsonplaceholder.typicode.com/posts/1"

updated_data = {
    "title": "Python Full Course API",
    "body": "Updated Python Content",
    "userId":1
}
# updated_data

response  = requests.put(url, json=updated_data)
print(response)
```

Note:

PUT = replace old data with new data every and each filed

### 10. PATCH --> Request(Update Data) ---> Minor data

```python
url = "https://jsonplaceholder.typicode.com/posts/1"

updated_data = {
    "title": "Python Full Course API",
    "userId":1
}
# updated_data

response  = requests.patch(url, json=updated_data)  # total column = 5, if you wanto to update 3 column then path, if 5 updated then put
print(response)
print(response.status_code)
```

### 11. DELETE Request



```python
url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.delete(url)  # Data deleted from server successfully
print(response.status_code)
```

### Sending Header in API

Headers used for:

* Authentication 
* content type
* Token

Example

```python
import requests
```

```python
url = "https://jsonplaceholder.typicode.com/posts"

headers = {
    "content-Type":"appliction/json"
}

response = requests.get(url,headers = headers)
print(response)
```

```python
url = "https://jsonplaceholder.typicode.com/posts"

headers = {
    "content-Type":"appliction/json"
}

response = requests.get(url,headers = headers)
print(response)
print(response.json())
```
