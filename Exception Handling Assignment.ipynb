```python
### **1. Basic Syntax**
#Write a Python program that takes a number from the user and prints its square. Use `try` and `except` to handle invalid (non-numeric) input.

def square_number():
    try:
        num = int(input("Enter a number: "))
    except ValueError:
        print("Invalid input! Enter a numeric value.")
    else:
        print(f"The square of {num} is {num ** 2}")
    finally:
        print("Program execution completed.")

square_number()
```

```python
##2. Multiple Except Blocks*

#Write code that divides two numbers entered by the user. Handle:

# * ValueError (invalid input)
# * ZeroDivisionError (denominator = 0)

def divide_numbers():
    try:
        numerator = int(input("Enter numerator: "))
        denominator = int(input("Enter denominator: "))
        
        div = numerator / denominator
    
    except ValueError:
         print("Invalid input! Please enter numeric values only.")
    
    except ZeroDivisionError:
         print("Error! Denominator cannot be zero.")
    
    else:
         print(f"Result: {numerator} / {denominator} = {div}")
    
    finally:
         print("Program execution completed.")


divide_numbers()
```

```python
### **10. Custom Exception**

#Create a custom exception `InvalidMarkError` and raise it when marks > 100 or < 0. Handle it using try-except.

  
# Function to validate marks
def check_marks(mark):
    try:
        if mark < 0 or mark > 100:
            raise InvalidMarkError(mark)
        else:
            print(f"Valid marks: {mark}")
    except Exception as e:
        print (e)
```

```python
### **11. Else With No Exception**

#Write a program where:

#* Try converts input to float
#* Except handles invalid input
#* Else prints “Conversion successful”

try:
    num = float(input("Enter a number: "))
except ValueError:
    print("Error: Enter a valid number.")
else:
    print("Conversion successful! You entered:", num)

```

```python
### **12. Handling KeyError**

#Write a program that accesses a dictionary key entered by the user and handles `KeyError`.

# Sample dictionary
student_marks = {
    "Mayuri": 85,
    "Shreya": 92,
    "Ankita": 78
}

try:
    name = input("Enter student name: ")
    
    print(f"{name} marks: {student_marks[name]}")

except KeyError:
    print(f"Error: No record found for '{name}'.")
```

```python
# **13. Handling IndexError**

#Ask the user for a list index and handle out-of-range cases.

# Sample list
numbers = [10, 20, 30, 40, 50]

try:
    index = int(input("Enter an index (0 to 4): "))
    print("Element at index", numbers[index])

except ValueError:
    print("Error: Enter a valid integer.")   # Handles non-integer input

except IndexError:
    print("Error: Index out of range. Valid range is 0 to 4 ")   # Handles out-of-range index
```

```python
### **14. Converting to Integer in a Loop**

#Loop until the user enters a valid integer. Use try-except to prevent program crash.

# Allow up to 5 attempts
for attempt in range(5):
    try:
        num = int(input("Enter an integer: "))
        print("You entered:", num)
        break  
    except ValueError:
        print("Error: Please enter a valid integer.")
else:
    print("Too many invalid attempts. Exiting...")
```
