# 🐍 Python Beginner Project Progression

These projects are arranged from **Beginner → Intermediate**.

| No. | Project | Concepts Covered | Hint |
|:---:|---------|------------------|------|
| 1 | 🧮 Calculator | Variables, Input/Output, Operators, Functions | Perform addition, subtraction, multiplication, and division based on user input. |
| 2 | 🎯 Number Guessing Game | Loops, Random Module, If-Else | Generate a random number and let the user guess until they find the correct number. |
| 3 | 📊 Student Marks Manager | List, Dictionary, Functions | Store student names and marks, then calculate total, average, highest, and lowest marks. |
| 4 | 📞 Phone Book | Dictionary, CRUD Operations | Add, search, update, delete, and display contact information. |
| 5 | 📝 Notes Manager | File Handling | Save, read, update, and delete notes from a text file. |
| 6 | 🏦 Bank Account System | OOP (Class & Object) | Implement deposit, withdraw, and balance check operations using classes. |
| 7 | 📚 Library Management System | OOP, File Handling | Manage books by adding, issuing, returning, and searching records. |
| 8 | 🌦️ Weather App (API) | Requests, JSON | Fetch and display live weather information from an API. |
| 9 | ✅ To-Do List (Tkinter) | GUI Programming | Create a desktop application to manage daily tasks. |
| 10 | 💬 Chat Application (Socket) | Socket Programming | Build a client-server application for real-time messaging. |
