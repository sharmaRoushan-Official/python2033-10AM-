4️⃣ `Self` Keyword
 * Self refers to the current object of the class
 > Why Self Keyword is Rquired?
   * To access instance variable and method 
   * Diffenciate instance and variable from local variable


```python

class Demo:

    # self is a reference to the current instance of the class, and is used to access variables that belong to the class.
    def show(self): # Method defined with self parameter

        print("Hello, this is a method in the Demo class.") 
d1 = Demo() # Creating an object d1 of the Demo class
d1.show() # Calling the show method using the object d1
```

```python
e1 = Demo() # Creating another object e1 of the Demo class
e1.show() # Calling the show method using the object e1 # self allows us to call the method for different objects of the same class, and it will work correctly for each object.
```

5️⃣ Constructor (_ _init_ _)

A `Constructor` initialize object data automatically when object is created.


```python
class Student: # Blueprint for creating student objects
    def __init__(self,name,age): # __init__ is a special method in Python classes, known as the constructor. It is automatically called when a new object of the class is created. The purpose of the __init__ method is to initialize the attributes of the object with the values provided as arguments during object creation.
        self.name = name # instance variable
        self.age = age # instance variable
s1 = Student("Ajay",30) # Creating an object s1 of the Student class and passing name and age as arguments
print(s1) # Printing the object s1, which will show its memory address
print(s1.name) # Accessing instance variable name through object s1
print(s1.age) # Accessing instance variable age through object s1
```

```python
y1 = Student("Yash",35)
print(y1) # Printing the object y1, which will show its memory address
print(y1.name) # Accessing instance variable name through object y1
print(y1.age) # Accessing instance variable age through object y1
```

```python
r1 = Student("Rohit",45)
print(r1)   # Printing the object r1, which will show its memory address (different from s1 and y1)
print(r1.name)
print(r1.age)
```

## 7. Types of Methods
 * 1. Instance Methods
   * Uses Self
   * Access intance variables
 * 2. Class Methods
   * Uses @classmethod
   * Uses cls
 * 3. Static Method
   * Uses @staticmethod
   * No self or cls

```python
class Student:
    collegeName = "Raisonic College,Nagpur" # class variable
    def __init__(self,name,age,address): # constructor
        self.name = name # instance name
        self.age = age  # instance age
        self.address = address # instance address
        
    print(f"College Name is:{collegeName}")  # calling class variables
    def studentDetails(self):  # method 
        print(f"Name of student is:{self.name}")
        print(f"Age of student is:{self.age}")
        print(f"Address of student is:{self.address}")


s1 = Student("Yash",22,"Nagpur") # s1 is object 
print(s1)  # calling object 
```

1. Instance Methods
* Uses Self
* Access intance variables


```python
class Student:
    def __init__(self,name,marks):
        self.name = name    # instance variable
        self.marks = marks # instance variables

    def show_details(self):  # Instance method
        print(f"Name:{self.name},Marks:{self.marks}")
# Usages
s1 = Student("Ravi",85)
s1.show_details()

s2 = Student("Priya",92)
s2.show_details()
```

## 2. Class Mehtod 
A mehtod that belongs to the class itself, not any specific object. 
It works on class-level data shares across all object.

#### Key Characteristics

 * Decorated with @classmethod
 * Taks cls as the first parameter(cls) = the class itself
 * Can access/modify class variable(shared by all object)
 * Can not access intance variables directly


```python
class Student:
    collegeName = "Raisonic College, Nagpur"

    def __init__(self,name):
        self.name = name
        
    @classmethod
    def show_college(cls):  # class method
        print(f"College Name:{cls.collegeName}")
    
    @classmethod
    def change_college(cls,new_name):  # class Method
        cls.collegeName = new_name

s1 = Student("Ravi")
s1.show_college()

s1.change_college("MIT, Pune")
s1.show_college()
```

## 3. Static Method

A method that is logically related to the class but does not need access to the class or any instance, 
It's essentially a plain funciton places inside a class for organizational purpose.

#### Key Characteristics
 * Decorated with @staticmethod
 * Takes No self or cls parameter
 * cannot access instance variable or class variables directly
 * used for utility/helper function

```python
class Student:
    
    @staticmethod
    def is_pass(marks):
        return marks>=75
        
    @staticmethod
    def grade(marks):
        if marks>=90:
            return "A+"
        elif marks>=75:
            return "A"
        elif marks>=60:
            return "B"
        else:
            return "C"

print(Student.is_pass(40))
print(Student.grade(40))
```

```python
print(Student.is_pass(95))
print(Student.grade(95))
```

```python
class Student:
    
    @staticmethod
    def is_pass(marks):
        print(marks>=75)  # value returned and terminated 
        
        if marks>=90:
            return "A+"
        elif marks>=75:
            return "A"
        elif marks>=60:
            return "B"
        else:
            return "C"

print(Student.is_pass(40))
print(Student.is_pass(95))
```
