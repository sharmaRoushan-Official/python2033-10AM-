## 2. Class Mehtod 
A mehtod that belongs to the class itself, not any specific object. 
It works on class-level data shares across all object.

#### Key Characteristics

 * Decorated with @classmethod
 * Taks cls as the first parameter(cls) = the class itself
 * Can access/modify class variable(shared by all object)
 * Can not access intance variables directly


```python
class Student:
    collegeName = "Raisonic College, Nagpur"

    def __init__(self,name):
        self.name = name
        
    @classmethod
    def show_college(cls):  # class method
        print(f"College Name:{cls.collegeName}")
    
    @classmethod
    def change_college(cls,new_name):  # class Method
        cls.collegeName = new_name

s1 = Student("Ravi")
s1.show_college()

s1.change_college("MIT, Pune")
s1.show_college()
```

```python
class Student:
    collegeName = "Raisonic College, Nagpur"  # class variable

    def __init__(self,name):
        self.name = name
        
    @classmethod
    def show_college(cls):
        print("College Name:",cls.collegeName)

s1 = Student("Rudra")
print(s1.name)
print(s1.collegeName)
s1.show_college()
```

## 3. Static Method

A method that is logically related to the class but does not need access to the class or any instance, 
It's essentially a plain funciton places inside a class for organizational purpose.

#### Key Characteristics
 * Decorated with @staticmethod
 * Takes No self or cls parameter
 * cannot access instance variable or class variables directly
 * used for utility/helper function

```python
class Student:
    
    @staticmethod
    def is_pass(marks):
        return marks>=75
        
    @staticmethod
    def grade(marks):
        if marks>=90:
            return "A+"
        elif marks>=75:
            return "A"
        elif marks>=60:
            return "B"
        else:
            return "C"

print(Student.is_pass(40))
print(Student.grade(40))
```

## 1. Single or Simple Inheritance

In __Single or Simple Inheritance__, a child class inherits from only __one parent__ class.


* 📌 Structure
* Parent -> Child

### Why It is Used

> Code reusability
> Child class can use parent class properties and method
> Easy to understand and maintan

### Real-life example

* Parent :- Vehicle
* Child :- Car
* A Car is a Vehicle

```python
class Vehicle:   # parent
    def start(self):
        print("Vehicle Started")
  
class Car(Vehicle):    # derived class or child child 
    def drive(self):
        print("Car is driving")

c1 = Car()
c1.drive()
c1.start()
```

```python
c2 = Car()
c2.start() # parent mehtod
c2.drive() # child method
```

## 2. Multiple Inheritance

In __Multiple Inheritance__, a child class inherits from more than one parent class.__

📌 Structure
* Parent1-------Parent2
*         \     /
*          child

### Why it is used

* When one class needs freature from multiple classes
* Helps to combine functionalities

### Real-life Example

* Parent1: Camera
* Parent2: Phone
* Child: Phone with camera

## Smart Phone is both a Phone and Camera with Internet connectivity

```python
class Phone:  # parent
    def call(self):
        print("Calling......")

        
class Camera:   # parent
    def click_photo(self):
        print("Photo Clicked")

class SmartPhone(Phone,Camera):  # derived Class : child
    def Internet(self):
        print('Conneting to Internet')

s1 = SmartPhone()
s1.Internet()
s1.call()
s1.click_photo()
```

## 3. Multilevel Inheritance

In __Multilevel Inheritance__, happes in levels.

A class inherits from a clss, and another class inherits from that child class.


📌 Structure

* Grandparent --> Parent --> Child

## Why it is used

* To represent Hierarchy in real world
* Step-by-step inheritance of feature

### Real-life eample

* Grandparent: Animal
* Parent: mammel
* Child: Dog

## Dog is Mammal, Mammal is Animal.

```python
class Animal:
    def birth(self):
        print("Reproduction")

class Mammal(Animal):
    def selfReproduction(self):
        print("Mammals naturally self reproduce")

class Dog(Mammal):
    def bark(self):
        print("Barking....")

d1 = Dog()
d1.birth()
d1.selfReproduction()
d1.bark()
```

## 4. Hierarchial Inheritance

In `Hierarchical Inheritance`, Multiple child classes inherits from single parent class.


📌 Structure

*          Parent
*      /      |       \
*  child1 --- child2 ----child3

### Why It is Used?

* One parent has common functionality
* Different children use and extend it different.


### Real-life Example

* Parent: Employee
* _Children_
  * Manager
  * Developer
  * Tester
  * HR

*** All are employees but have different roles

```python
class Employee: # parent
    def work(self):
        print("Employee are working")

    def extraWork(self):
        print("They all work overtime today")

class Manager(Employee):  # child1
    def manage(self):
        print("Manager are managing")

class Developer(Employee):  # child 2
    def code(self):
        print("Developer work coding")

    def tomorrowHoliday(self):
        print("Holiday Tomorrow")

class Tester(Employee):  # child 3
    def test(self):
        print("Tester Testing code")


m = Manager() # Employee + Manager
m.work()
m.manage()
m.extraWork()

d= Developer()
d.work()
d.code()
d.extraWork()
d.tomorrowHoliday()

t = Tester()
t.work()
t.test()
t.extraWork()
```

## 5. Hybrid Inheritance

Hybrid Inheritance is a combination of two or more inheritance types.

---
It may contains:

> Multiple + Multilevel

> Hierarchical + Multiple

📌 Structure 

*       A
*     /   \
*     B---C
*     \   /
*       D


Here:

* D and C inheritas from A(Hierarchical)
* D inherits from both B and C (Multiple)


### Why it is used

* When a system requires complex relationship
* Comman in large framework and enterprise projects.



```python
class A:  # parent
    def showA(self):
        print("Class A")

class B(A):  # child of A
    def showB(self):
        print("Class B")

class C(A):  # child of A
    def showC(self):
        print("Class C")

class D(B,C):  # B and C is the parent of D
    def showD(self):
        print("Class D")

obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
```

|Hybrid| Combination of Multiple types| A->B,C and B,C ->D|




## 2. Polymorphism

```python
class Animal:
    def sound(self): # sound is mehtod name and (value) is parameter: Signature
        print("Animal Makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog Barks")  # overriding


class Cat(Animal):
    def sound(self):
        print("Meaw")


d = Dog()
d.sound()

c = Cat()
c.sound()
```

👉 Method Overriding in Python 

Method Overridng means: 📌 A child class provides its own implementation of a method that is already defined in the parent class.

✅ Why Mehtod Overriding is Used?

1. To modify parent class behavior.
    > Child class may need different logic than parent.

2. To achieve Runtime polymorphism.
    > At runtime, Python decides which method to execute bases on the object.

3. To Implement specialization
    > Parent provides geneal method, child provide specific method.
    

### ✅ Rules of Method Overridng

* ✅Rule 1: Same method name
    > Parent and child mehtod names must be identical.
* ✅ Rule 2: Same parameters [recommeded]
    > The signature should match to avoid confusion.
* ✅ Rule 3: Happend only in inheritance
    > Overridng is possible only when there is parent-child class relationship.
* ✅ Rule 4: Child method is exexcute first
    > If child has the method, it overrides the parent method.
    

Exmple 1: Basic Method overriding

```python
class Animal:
    def sound(self):
        print("Animal makes sound...")

class Dog(Animal):
    def sound(self):
        print("Dog Barks")

d = Dog()
d.sound()
```

📌 Explanation:

Even sound() exists in parent class, child class version is executed.

### Exmaple 3: Method Overrindg + Super()


## What is super()?

super() is used to call the parent class method inside the child class.


📌 Use When:

* You want parent functionality also
* plus some extra child functionality.

```python
class Employee:
    def work(self):
        print("Employee is working")
        print("This is parent method.")

class Manager(Employee):
    def work(self):
        super().work()  # calling parent method
        print("Manager is managing the team.")
        x = int(input("Enter first number: "))
        y = int(input("Enter second number: "))
        sum1 = x+y
        print(f"The sum of two numbers {x} and {y} is {sum1}")
        


m = Manager()
m.work()
```

```python
class Employee:
    def __init__(self):
        self.__salary = 0

    @property
    def salary(self):
        return self.__salary

    def salary(self, amount):
        if amount>0:
            self.__salary = amount
        else:
            print("Invalid Salary")


emp = Employee()
emp.salary = 40000
print("Salary:",emp.salary)
```

```python
### Polymorphism 

# --------------
# Overriding:
# Overloading:

class Shape:  # parent
    def area(self):
        return "Area Formula depend on shape"

class Circle(Shape):
    def area(self):
        return "Area: πr²"

c1 = Circle()
c1.area()
```

```python
# Overloading: *args


def sumValue(*args):
    return f"{sum(args)}"

sumValue(10)
```

```python
sumValue(10,20,30,)
```

```python
class Student:
    def __init__(self,name):
        self.name = name
        
s1 = Student("Mohammad")
s1.name
```

```python
class Student2(Student):
    def __init__(self,name,age):
        self.age = age
        super().__init__(name)

s2 = Student2("Yash",25)
print(s2.name)
print(s2.age)
```

```python
class Student3(Student2):
    def __init__(self,name,age,address):
        self.address = address
        super().__init__(name,age)

s3 = Student3("Mohammad",32,"Noida")
s3.name
```

```python
# Method Overloading 

class Calculator:
    # method 
    def add(self,a,b=0,c=0):  # parameter fix max-3
        return a+b+c
obj = Calculator()
print(obj.add(10))
print(obj.add(10,20))
print(obj.add(10,20,30))
print(obj.add(10,20,30,40))
```

```python
class Calculator:
    # method 
    def add(self,*a):
        return sum(a)
obj = Calculator()
print(obj.add(10))
print(obj.add(10,20))
print(obj.add(10,20,30))
print(obj.add(10,20,30,40))
print(obj.add(10,20,30,40,50,60,))
```

```python
# Method Ovaerlading 
class Calculator:
    # method 
    def add(self,*a):
        total = 0
        for num in a:
            total+=num
        return f"Sum of the value of given {a} is: {total}"
            
obj = Calculator()
print(obj.add(10))
print(obj.add(10,20))
print(obj.add(10,20,30))
print(obj.add(10,20,30,40))
print(obj.add(10,20,30,40,50,60,))
```
