### 1. What is a File in Python?

A `file` is used to store data permanently on a disk.Python allows us to create, read, write and modify files.

Example:

* .txt ---> Text file
* .csv ---> Data file
* .json --> Structured data file
* .log  ---> logs file
* .pdf  ---> Document File
* .xlsx  ---> sheet file
* .docx ----> document file

Python provides a built-in function:

open()

## 2. Syntax of Opening a File

file_object = open("fileName","mode")


Parameters:

|Prameter|Description|
|----|----|
|fileName|Name of the file|
|mode|Operation to perform|

## 3. Files Modes

|Mode|Description|
|---|---|
|`r`|Read file(default)|
|`w`|Write file(overwrite)|
|`a`|Append Data|
|`x`|Create file|
|`rb`|Read binary|
|`wb`|Write binary|

```python
# Data Abstraction

# import module ABC, abstractmethod

from abc import ABC
```

```python
help(ABC)
```

```python
from abc import ABC, abstractmethod
class Bank(ABC):
    @abstractmethod
    def security(self):
        pass
```

```python
class SBI(Bank):
    def customerInfo(self):
        print("Customer Info")

    def security(self):
        return "This is SBI Security"

b1 = SBI()
b1.customerInfo()
b1.security()
```

```python
# Write a function display_info(obj) that calls obj.info(). 
# Pass objects of two different classes having an info() method to demonstrate polymorphism.


class Animal:
    def display_info(self,domain,Kingdom):
        pass

class Cat(Animal):
    def display_info(self,domain,kingdom):
        print(f"Cat domain is:{domain} \nCat Kingdom is:{kingdom}")

c1 = Cat()
c1.display_info("Yukaira","Animalia")
```
