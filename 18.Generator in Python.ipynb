### 18. Generator in Python

### What is a Generator in Python?

A `Generator` is a special type of function that produces values one at a time, instead of producing all values at once.


A generator doest not stores all results in memory like a list.

Instead, it generates the next value only when needed.

This called:

✅ Lazy Evaluation

✅ Memory Efficient Iteration

### ✅ Normal Funciton vs Generator Function

```python
def normal():
    return 10

print(normal())
```

### Generator Function

A generator uses yield instead of return

It pauses execution and resume later

```python
def gen():
    yield 10
print(gen())
```

```python
print(next(gen()))
```

The generator function return a generator object, not the value

✅ What does yield do?

The yiled keyword:
* Produces a value
* Pauses the function state
* Saves local variables
* Resumes from the same point when called again

```python
def my_generator():
    yield 1
    yield 2
    yield 3
    yield 4
    yield 5
g = my_generator()
print(g)  # generator object 
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
# print(next(g)) #StopIteration:
```

## Generator Execution Flow

```python
def test():
    print("Start")
    yield 100
    print("Middle")
    yield 200
    print("End")
    yield 300
```

```python
obj = test()
print(obj)
```

```python
# next() keyword 

print(next(obj))
```

```python
print(next(obj))
```

```python
print(next(obj))
```

### Explain

* first next() runs until first yield.
* second next() resume after first yield.
* third next() resumes after second yiled.


So, generator works like pause-resume machine.

### Why Generators are Memory Efficient?

* List Example(consumes memory)

```python
def list_numbers(n):
    return [i for i in range(n)]

lst = list_numbers(1000000)
print(len(lst))
```

This creates 1 million numbers in memory

Generator Example(No memory waste)

```python
def gen_numbers(n):
    for i in range(n):
        yield i

gen = gen_numbers(1000000)
print(gen)
```

```python
def list_numbers(n):
    return [i for i in range(n)]

lst = list_numbers(100000000)
print(len(lst))
```

```python
def gen_numbers(n):
    for i in range(n):
        yield i

gen = gen_numbers(1000000000000000)
print(gen)
```

```python
def list_numbers(n):
    return [i for i in range(n)]

lst = list_numbers(10000000000)
print(len(lst))
```

### ✅ yield vs return

|Keyword|Behavior|
|-----|-----|
|return|Ends function permanently[terminate function]|
|yield|Pause funciton and resume later(next)|

### Generator Function with send()

Generators can accept input dynamically using send()

```python
def mygen():
    x = yield "Hello"
    yield x

g = mygen()
print(next(g))  # start generator
```

```python
print(g.send(50))  # Send value 
```

first next() start generator

yield "Hello" return "Hello"

send(50) assigns 50 to x

### Generator with throw()

You can raise exception inside generator

```python
def gen():
    try:
        yield 1
        yield 2
    except ValueError:
        yield "ValueError Occurred"
g = gen()
print(next(g))
```

```python
print(g.throw(ValueError))
```

### Generator with close()

Stops generator permananetly


```python
def gen():
    yield 1
    yield 2
    yield 3
g = gen()
print(next(g))
```

```python
g.close()
```

```python
print(next(g))
```

```python
print(next(g))
```

Iterable vs iterator()

|Term|Meaning|
|----|---|
|Iterable|Object that can be looped(list, tuple, string)|
|Iterator| Object that producees value using __next__()|

Generator are iterators automatically.


A generator Implements:

* __iter__() ---> return itself
* __next__() ---> gives next value

```python
g=(x for x in range(3))
print(iter(g) == g)
```

```python
for x in range(3):
    print(x)
```
