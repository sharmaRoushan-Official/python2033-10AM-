```python
#1)1. Basic Syntax*

#Write a Python program that takes a number from the user and prints its square. 
# Use try and except to handle invalid (non-numeric) input.

try:
    a = int(input("Enter Base Value: "))
    b = int(input("Enter Exponential Value: "))
    assert a<=10,"Invalid Number,Enter value less than 10."
    res = a**b
except ValueError:
    print("This is a value error")
else:
    print(f"Square of given numbers base {a}, Exponent:{b}-Result:{res}")
```

```python
#2) *2. Multiple Except Blocks*
    #Write code that divides two numbers entered by the user. Handle:
    #* ValueError (invalid input)
      #* ZeroDivisionError (denominator = 0)

try:
    a = int(input("Enter any number"))
    b = int(input("Enter any number"))
    print(a/b)
except ZeroDivisionError:
    print("division by zero")
except ValueError:
    print("This is a value error")
```
