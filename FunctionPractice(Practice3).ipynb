```python
## 1. Wap to find the are of square and thier perimeter

# PascalCase- class or object
# snake_case- variable
# camelCase- function

def areaSquare(side):
    area  = side*side
    print("Area of Square:",area,"cm²")
    perimeter = 4*side
    print(f"Perimeter of square is {perimeter} cm")
areaSquare(10)
```

```python
def areaSquare():
    
    side = int(input("Enter side of Square:"))
    if side>0:
        area  = side*side
        print("Area of Square:",area,"cm²")
        perimeter = 4*side
        print(f"Perimeter of square is {perimeter} cm")
    else:
        print("Side of Square cannot be in Negative")
areaSquare()
```
